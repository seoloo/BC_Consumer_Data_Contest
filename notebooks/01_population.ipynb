```python
import csv
import pandas as pd

with open(future_path, "r", encoding="utf-8-sig", newline="") as f:
    rows = list(csv.reader(f))

future = pd.DataFrame(
    rows[1:],       # 실제 데이터
    columns=rows[0] # 첫 줄을 컬럼명으로 사용
)

future.head()
```

```python
# 컬럼 이름 앞뒤 공백 정리
future.columns = future.columns.str.strip()

print("자치구:")
print(future["자치구별"].unique())

print("\n성별:")
print(future["성별"].unique())

print("\n연령:")
print(future["연령별"].unique()[:40])
```

```python
# 사용할 5세 단위 연령만 선택
age_groups_5 = [
    "0~4세",
    "5~9세",
    "10~14세",
    "15~19세",
    "20~24세",
    "25~29세",
    "30~34세",
    "35~39세",
    "40~44세",
    "45~49세",
    "50~54세",
    "55~59세",
    "60~64세",
    "65~69세",
    "70~74세",
    "75~79세",
    "80~84세",
    "85~89세",
    "90세 이상"
]

future_5 = future[
    (future["성별"] == "계") &
    (future["연령별"].isin(age_groups_5))
].copy()

future_5.head()
```

```python
# 서울 전체 합계 제외
future_5 = future_5[future_5["자치구별"] != "합계"].copy()

# 5세 단위를 우리가 쓸 연령대로 변환
age_map = {
    "0~4세": "20대미만",
    "5~9세": "20대미만",
    "10~14세": "20대미만",
    "15~19세": "20대미만",

    "20~24세": "20대",
    "25~29세": "20대",

    "30~34세": "30대",
    "35~39세": "30대",

    "40~44세": "40대",
    "45~49세": "40대",

    "50~54세": "50대",
    "55~59세": "50대",

    "60~64세": "60대이상",
    "65~69세": "60대이상",
    "70~74세": "60대이상",
    "75~79세": "60대이상",
    "80~84세": "60대이상",
    "85~89세": "60대이상",
    "90세 이상": "60대이상"
}

future_5["연령대"] = future_5["연령별"].map(age_map)

future_5[["자치구별", "연령별", "연령대"]].head(15)
```

```python
# 2022~2042년 컬럼
year_cols = [f"{year} 년" for year in range(2022, 2043)]

# 연도별 값을 숫자로 변환
for col in year_cols:
    future_5[col] = pd.to_numeric(future_5[col], errors="coerce").fillna(0)

# 자치구 + 연령대별로 합치기
future_age = (
    future_5
    .groupby(["자치구별", "연령대"])[year_cols]
    .sum()
    .reset_index()
)

future_age.head(10)
```

```python
future_change = future_age[
    ["자치구별", "연령대", "2026 년", "2030 년", "2035 년", "2040 년"]
].copy()

for year in [2030, 2035, 2040]:
    future_change[f"{year} 증감률"] = (
        (future_change[f"{year} 년"] - future_change["2026 년"])
        / future_change["2026 년"]
        * 100
    ).round(1)

future_change.head(12)
```

```python
future_change.to_csv(
    "../data/processed/seoul_future_population_change.csv",
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료")
```

```python
import pandas as pd

busan_path = "../data/raw/busan_future_population_2015_2040.xlsx"

busan = pd.read_excel(
    busan_path,
    sheet_name=0,
    header=0
)

print(busan.columns.tolist())

busan.head(10)
```

```python
# 시군구명이 빈 행은 위의 값을 이어받기
busan["시군구별"] = busan["시군구별"].ffill()

# 성별도 같은 방식
busan["성별"] = busan["성별"].ffill()

print(busan["시군구별"].drop_duplicates().tolist())
```

```python
# 1. 성별 전체 + 5세 구간 소계만 사용
busan_5age = busan[
    (busan["성별"] == "계") &
    (busan["연령별(2)"] == "소계") &
    (busan["시군구별"] != "부산광역시")
].copy()

# 연령 문자열 정리
busan_5age["연령구간"] = (
    busan_5age["연령별(1)"]
    .astype(str)
    .str.replace(" ", "", regex=False)
)

# 확인
busan_5age[
    ["시군구별", "연령구간", 2026, 2030, 2035, 2040]
].head(20)
```

```python
# 2. 부산 5세 연령구간 → BC 연령대 기준으로 변환

def busan_age_group(age_text):
    
    # 90세 이상, 100세 이상 같은 경우
    if "이상" in age_text:
        return "60대이상"

    # 0-4세 → 앞 숫자 0
    start_age = int(age_text.split("-")[0])

    if start_age < 20:
        return "20대이하"
    elif start_age < 30:
        return "20대"
    elif start_age < 40:
        return "30대"
    elif start_age < 50:
        return "40대"
    elif start_age < 60:
        return "50대"
    else:
        return "60대이상"


busan_5age["연령대"] = busan_5age["연령구간"].apply(
    busan_age_group
)

# 연도별 인구 숫자형 변환
years = [2026, 2030, 2035, 2040]

for year in years:
    busan_5age[year] = pd.to_numeric(
        busan_5age[year],
        errors="coerce"
    )

# 구군 × 연령대별 합계
busan_future = (
    busan_5age
    .groupby(
        ["시군구별", "연령대"],
        as_index=False
    )[years]
    .sum()
)

busan_future.head(20)
```

```python
# 2. 부산 5세 연령구간 → BC 연령대 기준으로 변환

# '계' 행 제거
busan_5age = busan_5age[
    busan_5age["연령구간"] != "계"
].copy()


def busan_age_group(age_text):

    # 예: 90세이상, 100세이상
    if "이상" in age_text:
        return "60대이상"

    # 예: 0-4세, 5-9세
    start_age = int(
        age_text.split("-")[0]
        .replace("세", "")
    )

    if start_age < 20:
        return "20대이하"
    elif start_age < 30:
        return "20대"
    elif start_age < 40:
        return "30대"
    elif start_age < 50:
        return "40대"
    elif start_age < 60:
        return "50대"
    else:
        return "60대이상"


busan_5age["연령대"] = (
    busan_5age["연령구간"]
    .apply(busan_age_group)
)

# 연도별 숫자형 변환
years = [2026, 2030, 2035, 2040]

for year in years:
    busan_5age[year] = pd.to_numeric(
        busan_5age[year],
        errors="coerce"
    )

# 구군 × 연령대 합계
busan_future = (
    busan_5age
    .groupby(
        ["시군구별", "연령대"],
        as_index=False
    )[years]
    .sum()
)

busan_future.head(20)
```

```python
# 3. 2026년 대비 미래 인구 증감률 계산

for year in [2030, 2035, 2040]:
    busan_future[f"{year} 증감률"] = (
        (
            busan_future[year]
            - busan_future[2026]
        )
        / busan_future[2026]
        * 100
    ).round(1)

# 컬럼명 서울 데이터와 맞추기
busan_future = busan_future.rename(
    columns={"시군구별": "시군구"}
)

busan_future.head(20)
```

```python
busan_future.to_csv(
    "../data/processed/busan_future_population_change.csv",
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료")
```

```python
import pandas as pd
import re

all_flow = []

for file in csv_files:

    # 파일 읽기
    try:
        temp = pd.read_csv(file, encoding="utf-8-sig")
    except UnicodeDecodeError:
        temp = pd.read_csv(file, encoding="cp949")

    # 파일명에서 기준년월 추출
    month = re.search(r"20\d{4}", file.stem).group()

    temp["기준년월"] = month

    # 시군명
    temp["시군"] = temp["CTY_NM"].replace({
        "연천시": "연천군"
    })

    all_flow.append(temp)

flow = pd.concat(all_flow, ignore_index=True)

# 숫자형 연령 컬럼
age_cols = [
    "M_10_CNT", "M_15_CNT",
    "M_20_CNT", "M_25_CNT",
    "M_30_CNT", "M_35_CNT",
    "M_40_CNT", "M_45_CNT",
    "M_50_CNT", "M_55_CNT",
    "M_60_CNT", "M_65_CNT", "M_70_CNT",
    "F_10_CNT", "F_15_CNT",
    "F_20_CNT", "F_25_CNT",
    "F_30_CNT", "F_35_CNT",
    "F_40_CNT", "F_45_CNT",
    "F_50_CNT", "F_55_CNT",
    "F_60_CNT", "F_65_CNT", "F_70_CNT"
]

flow[age_cols] = (
    flow[age_cols]
    .apply(pd.to_numeric, errors="coerce")
    .fillna(0)
)

# BC 연령대와 동일하게 변환
flow["20대이하"] = flow[
    ["M_10_CNT", "M_15_CNT", "F_10_CNT", "F_15_CNT"]
].sum(axis=1)

flow["20대"] = flow[
    ["M_20_CNT", "M_25_CNT", "F_20_CNT", "F_25_CNT"]
].sum(axis=1)

flow["30대"] = flow[
    ["M_30_CNT", "M_35_CNT", "F_30_CNT", "F_35_CNT"]
].sum(axis=1)

flow["40대"] = flow[
    ["M_40_CNT", "M_45_CNT", "F_40_CNT", "F_45_CNT"]
].sum(axis=1)

flow["50대"] = flow[
    ["M_50_CNT", "M_55_CNT", "F_50_CNT", "F_55_CNT"]
].sum(axis=1)

flow["60대이상"] = flow[
    [
        "M_60_CNT", "M_65_CNT", "M_70_CNT",
        "F_60_CNT", "F_65_CNT", "F_70_CNT"
    ]
].sum(axis=1)

print("전체 행 수:", len(flow))
print("기준년월:", sorted(flow["기준년월"].unique()))
print("시군 수:", flow["시군"].nunique())
print("시군:", sorted(flow["시군"].unique()))
```

```python
# 구 단위를 부모 시 단위로 통일
flow["시군"] = (
    flow["CTY_NM"]
    .astype(str)
    .str.split()
    .str[0]
    .replace({
        "연천시": "연천군"
    })
)

print("시군 수:", flow["시군"].nunique())
print("시군 목록:")
print(sorted(flow["시군"].unique()))
```

```python
age_groups = [
    "20대이하",
    "20대",
    "30대",
    "40대",
    "50대",
    "60대이상"
]

# 1. 시군 × 월 기준으로 유동인구 합산
flow_monthly = (
    flow
    .groupby(["기준년월", "시군"], as_index=False)[age_groups]
    .sum()
)

# 2. 월별 전체 유동인구
flow_monthly["전체유동인구"] = (
    flow_monthly[age_groups].sum(axis=1)
)

# 3. 각 연령대가 해당 시군 월 전체에서 차지하는 비중
for age in age_groups:
    flow_monthly[f"{age}_비중"] = (
        flow_monthly[age]
        / flow_monthly["전체유동인구"]
        * 100
    )

print("행 수:", len(flow_monthly))
print("월:", sorted(flow_monthly["기준년월"].unique()))
print("시군 수:", flow_monthly["시군"].nunique())

display(
    flow_monthly[
        ["기준년월", "시군"]
        + [f"{age}_비중" for age in age_groups]
    ].head(10)
)
```

```python
share_cols = [
    "20대이하_비중",
    "20대_비중",
    "30대_비중",
    "40대_비중",
    "50대_비중",
    "60대이상_비중"
]

# 시군별 2026년 1~3월 평균 연령비중
flow_age_avg = (
    flow_monthly
    .groupby("시군", as_index=False)[share_cols]
    .mean()
)

# 보기 좋게 소수 둘째 자리
flow_age_avg[share_cols] = flow_age_avg[share_cols].round(2)

# 합계 확인
flow_age_avg["비중합계"] = (
    flow_age_avg[share_cols].sum(axis=1)
).round(2)

print("시군 수:", len(flow_age_avg))

display(flow_age_avg)
```

```python
# 비교 연령대
ages = ["20대", "30대", "40대", "50대", "60대이상"]

# 경기 BC만 추출
bc_gg = bc_age[
    (bc_age["시도"] == "경기도") &
    (bc_age["연령대"].isin(ages))
].copy()

# BC의 구 단위를 부모 시 단위로 통일
def to_parent_city(name):
    parts = str(name).split()
    if len(parts) >= 2 and parts[0].endswith("시"):
        return parts[0]
    return name

bc_gg["시군"] = bc_gg["시군구"].apply(to_parent_city)

# 유동인구가 있는 16개 시군만
available_cities = compare["시군"].unique()

bc_gg = bc_gg[
    bc_gg["시군"].isin(available_cities)
].copy()

# 시군 × 업종 × 연령으로 다시 합산
bc_gg = (
    bc_gg
    .groupby(
        ["시군", "업종코드", "업종명", "연령대"],
        as_index=False
    )["이용금액"]
    .sum()
)

# 20대 이상 안에서 소비비중 재계산
bc_gg["20대이상_업종총액"] = (
    bc_gg
    .groupby(["시군", "업종코드", "업종명"])["이용금액"]
    .transform("sum")
)

bc_gg["소비비중_20plus"] = (
    bc_gg["이용금액"]
    / bc_gg["20대이상_업종총액"]
)

print("시군 수:", bc_gg["시군"].nunique())
print("업종 수:", bc_gg["업종명"].nunique())

display(bc_gg.head(10))
```

```python
# 1. 유동인구 보정계수를 long 형태로 변환
adjust_long = compare[
    [
        "시군",
        "20대_보정계수",
        "30대_보정계수",
        "40대_보정계수",
        "50대_보정계수",
        "60대이상_보정계수"
    ]
].melt(
    id_vars="시군",
    var_name="연령대",
    value_name="유동보정계수"
)

# '20대_보정계수' -> '20대'
adjust_long["연령대"] = (
    adjust_long["연령대"]
    .str.replace("_보정계수", "", regex=False)
)

# 2. 2040 장래인구 증감률 붙이기
future_2040 = future_gg[
    ["시군", "연령대", "2040 증감률"]
].copy()

# 3. BC 소비구조 + 유동보정 + 장래인구 결합
gg_robust = (
    bc_gg
    .merge(
        adjust_long,
        on=["시군", "연령대"],
        how="left"
    )
    .merge(
        future_2040,
        on=["시군", "연령대"],
        how="left"
    )
)

print("행 수:", len(gg_robust))
print("시군 수:", gg_robust["시군"].nunique())
print("업종 수:", gg_robust["업종명"].nunique())

print(
    "유동보정계수 결측:",
    gg_robust["유동보정계수"].isna().sum()
)

print(
    "2040 증감률 결측:",
    gg_robust["2040 증감률"].isna().sum()
)

display(
    gg_robust[
        [
            "시군",
            "업종명",
            "연령대",
            "소비비중_20plus",
            "유동보정계수",
            "2040 증감률"
        ]
    ].head(10)
)
```

```python
# 1. 유동인구 보정계수를 BC 소비비중에 적용
gg_robust["보정전_가중치"] = (
    gg_robust["소비비중_20plus"]
)

gg_robust["보정가중치_raw"] = (
    gg_robust["소비비중_20plus"]
    * gg_robust["유동보정계수"]
)

# 2. 시군 × 업종 안에서 다시 합이 1이 되도록 정규화
gg_robust["보정가중치"] = (
    gg_robust["보정가중치_raw"]
    / gg_robust
        .groupby(["시군", "업종코드"])["보정가중치_raw"]
        .transform("sum")
)

# 3. 기존 Stress / 유동인구 보정 Stress 계산
gg_robust["기존Stress_연령"] = (
    gg_robust["보정전_가중치"]
    * gg_robust["2040 증감률"]
)

gg_robust["보정Stress_연령"] = (
    gg_robust["보정가중치"]
    * gg_robust["2040 증감률"]
)

# 4. 시군 × 업종별로 합산
gg_stress_compare = (
    gg_robust
    .groupby(
        ["시군", "업종코드", "업종명"],
        as_index=False
    )
    .agg(
        기존Stress=("기존Stress_연령", "sum"),
        보정Stress=("보정Stress_연령", "sum")
    )
)

# 5. 변화량과 위험/안정 분류
gg_stress_compare["Stress변화"] = (
    gg_stress_compare["보정Stress"]
    - gg_stress_compare["기존Stress"]
)

gg_stress_compare["기존분류"] = (
    gg_stress_compare["기존Stress"] >= 0
).map({
    True: "안정",
    False: "위험"
})

gg_stress_compare["보정분류"] = (
    gg_stress_compare["보정Stress"] >= 0
).map({
    True: "안정",
    False: "위험"
})

gg_stress_compare["분류일치"] = (
    gg_stress_compare["기존분류"]
    == gg_stress_compare["보정분류"]
)

print("사례 수:", len(gg_stress_compare))

print(
    "분류 일치율:",
    round(gg_stress_compare["분류일치"].mean() * 100, 1),
    "%"
)

print(
    "분류 변경:",
    (~gg_stress_compare["분류일치"]).sum(),
    "건"
)

display(
    gg_stress_compare.sort_values(
        "Stress변화"
    ).head(10)
)
```

```python
# 생활인구 구조로 BC 소비비중 재가중
seoul_robust["보정가중치_raw"] = (
    seoul_robust["소비비중"]
    * seoul_robust["생활보정계수"]
)

seoul_robust["보정소비비중"] = (
    seoul_robust["보정가중치_raw"]
    / seoul_robust
      .groupby(["시군구", "업종코드"])["보정가중치_raw"]
      .transform("sum")
)

# 기존 / 보정 Stress
seoul_robust["기존Stress_연령"] = (
    seoul_robust["소비비중"]
    * seoul_robust["2040 증감률"]
)

seoul_robust["보정Stress_연령"] = (
    seoul_robust["보정소비비중"]
    * seoul_robust["2040 증감률"]
)

seoul_stress_compare = (
    seoul_robust
    .groupby(
        ["시군구", "업종코드", "업종명"],
        as_index=False
    )
    .agg(
        기존Stress=("기존Stress_연령", "sum"),
        보정Stress=("보정Stress_연령", "sum")
    )
)

seoul_stress_compare["Stress변화"] = (
    seoul_stress_compare["보정Stress"]
    - seoul_stress_compare["기존Stress"]
)

print("사례 수:", len(seoul_stress_compare))

display(seoul_stress_compare.head(10))
```

```python
from scipy.stats import spearmanr
import pandas as pd

# 1. 기존 / 보정 위험·안정 분류
seoul_stress_compare["기존분류"] = (
    seoul_stress_compare["기존Stress"] >= 0
).map({
    True: "안정",
    False: "위험"
})

seoul_stress_compare["보정분류"] = (
    seoul_stress_compare["보정Stress"] >= 0
).map({
    True: "안정",
    False: "위험"
})

seoul_stress_compare["분류일치"] = (
    seoul_stress_compare["기존분류"]
    == seoul_stress_compare["보정분류"]
)

# 2. 분류 일치율
agreement = (
    seoul_stress_compare["분류일치"].mean() * 100
)

changed_n = (
    ~seoul_stress_compare["분류일치"]
).sum()

print("사례 수:", len(seoul_stress_compare))
print("분류 일치율:", round(agreement, 1), "%")
print("분류 변경:", changed_n, "건")

# 3. 변경 방향
change_table = pd.crosstab(
    seoul_stress_compare["기존분류"],
    seoul_stress_compare["보정분류"]
)

print("\n분류 변화표")
display(change_table)

# 4. Stress 위험순위 상관
rho, p = spearmanr(
    seoul_stress_compare["기존Stress"],
    seoul_stress_compare["보정Stress"]
)

print("Spearman rho:", round(rho, 3))
print("p-value:", p)

# 5. 변경 사례
seoul_changed = seoul_stress_compare[
    seoul_stress_compare["분류일치"] == False
].copy()

seoul_changed["기존Stress절대값"] = (
    seoul_changed["기존Stress"].abs()
)

display(
    seoul_changed.sort_values("기존Stress절대값")[
        [
            "시군구",
            "업종명",
            "기존Stress",
            "보정Stress",
            "Stress변화",
            "기존분류",
            "보정분류",
            "기존Stress절대값"
        ]
    ]
)
```

```python
# -------------------------
# 1. Q75 기준 서울 사례
# -------------------------

q75_seoul = q75_base[
    q75_base["시도"] == "서울특별시"
][
    ["시군구", "업종코드", "업종명"]
].drop_duplicates()

seoul_q75_robust = (
    q75_seoul
    .merge(
        seoul_stress_compare,
        on=["시군구", "업종코드", "업종명"],
        how="inner"
    )
)

print("서울 Q75 강건성 사례 수:", len(seoul_q75_robust))


# -------------------------
# 2. Q75 기준 경기 사례
# -------------------------

q75_gg = q75_base[
    q75_base["시도"] == "경기도"
].copy()

# 경기 robust 쪽은 컬럼명이 '시군'
q75_gg["시군"] = q75_gg["시군구"].apply(to_parent_city)

q75_gg = q75_gg[
    ["시군", "업종코드", "업종명"]
].drop_duplicates()

gg_q75_robust = (
    q75_gg
    .merge(
        gg_stress_compare,
        on=["시군", "업종코드", "업종명"],
        how="inner"
    )
)

print("경기 Q75 강건성 사례 수:", len(gg_q75_robust))
```

```python
from scipy.stats import spearmanr

def robustness_summary(df, region):
    
    # 분류가 아직 없으면 생성
    df = df.copy()

    df["기존분류"] = (
        df["기존Stress"] >= 0
    ).map({
        True: "안정",
        False: "위험"
    })

    df["보정분류"] = (
        df["보정Stress"] >= 0
    ).map({
        True: "안정",
        False: "위험"
    })

    df["분류일치"] = (
        df["기존분류"] == df["보정분류"]
    )

    agreement = df["분류일치"].mean() * 100
    changed = (~df["분류일치"]).sum()

    rho, p = spearmanr(
        df["기존Stress"],
        df["보정Stress"]
    )

    return {
        "지역": region,
        "사례수": len(df),
        "분류일치율": round(agreement, 1),
        "변경건수": int(changed),
        "Spearman_rho": round(rho, 3),
        "p_value": p
    }


seoul_summary = robustness_summary(
    seoul_q75_robust,
    "서울"
)

gg_summary = robustness_summary(
    gg_q75_robust,
    "경기"
)

pd.DataFrame([
    seoul_summary,
    gg_summary
])
```
